## **Twitter API**

```python
# Twitter API Keys
API_Key = os.getenv('TwitterAPI_Key')
API_Secret_Key = os.getenv('TwitterAPI_Secret_Key')
Bearer_Token = 'KEY'
Access_Token = os.getenv('TwitterAPI_Access_Token')
Access_Token_Secret = os.getenv('TwitterAPI_Access_Token_Secret')

# Authenticate to Twitter
auth = tweepy.OAuthHandler(API_Key, API_Secret_Key)
auth.set_access_token(Access_Token,Access_Token_Secret)
api = tweepy.API(auth)
```
